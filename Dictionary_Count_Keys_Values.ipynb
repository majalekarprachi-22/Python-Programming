```python

student = {
    "Prachi": 85,
    "Riya": 90,
    "Nidhi": 78,
    "Sneha": 95
}

total_keys = len(student.keys())

total_values = len(student.values())

print("Student Dictionary:", student)
print("Total Keys:", total_keys)
print("Total Values:", total_values)
```

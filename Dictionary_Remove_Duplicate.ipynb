```python

student = {
    "Prachi": 85,
    "Riya": 90,
    "Nidhi": 85,
    "Sneha": 95,
    "Pooja": 90
}

unique_student = {}

for key, value in student.items():
    if value not in unique_student.values():
        unique_student[key] = value

print("Original Dictionary:", student)
print("Dictionary After Removing Duplicate Values:", unique_student)
```

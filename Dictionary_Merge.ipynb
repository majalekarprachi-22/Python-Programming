```python

dict1 = {
    "name": "Prachi",
    "age": 20
}

dict2 = {
    "city": "Kolhapur",
    "marks": 85
}

dict1.update(dict2)

print("First Dictionary:", dict1)
print("Second Dictionary:", dict2)
print("Merged Dictionary:", dict1)
```

```python
student = {
    "Prachi": 85,
    "Riya": 90,
    "Nidhi": 78,
    "Sneha": 95
}

name = input("Enter student name to search: ")

if name in student:
    print(name, "is present in dictionary")
    print("Marks:", student[name])
else:
    print(name, "is not present in dictionary")
```

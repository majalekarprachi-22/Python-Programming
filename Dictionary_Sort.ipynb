```python

marks = {
    "Prachi": 85,
    "Riya": 90,
    "Nidhi": 78,
    "Sneha": 95
}

sorted_marks = dict(sorted(marks.items(), key=lambda x: x[1]))

print("Original Dictionary:", marks)
print("Sorted Dictionary:", sorted_marks)
```

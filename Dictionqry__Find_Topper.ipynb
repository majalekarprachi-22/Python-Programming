```python

students = {
    "Prachi": 85,
    "Riya": 90,
    "Nidhi": 78,
    "Sneha": 95
}

topper = max(students, key=students.get)

print("Student Marks:", students)
print("Topper:", topper)
print("Highest Marks:", students[topper])
```

```python

student = {
    "name": "Prachi",
    "age": 20,
    "marks": 85
}

print("Original Dictionary:", student)

student["city"] = "Kolhapur"
print("After Adding:", student)

student["marks"] = 90
print("After Updating:", student)

del student["age"]
print("After Deleting:", student)
```

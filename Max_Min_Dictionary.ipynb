```python

marks = {
    "Prachi": 85,
    "Riya": 90,
    "Nidhi": 78,
    "Sneha": 95
}

maximum = max(marks.values())
minimum = min(marks.values())

print("Student Marks:", marks)
print("Maximum Marks:", maximum)
print("Minimum Marks:", minimum)
```
